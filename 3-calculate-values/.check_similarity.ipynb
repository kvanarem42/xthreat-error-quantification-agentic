```python
import pandas as pd
import numpy as np
```

```python
df_old = pd.read_csv('bootstrap_errors_normal_xthreat.csv')
df_new = pd.read_csv('bootstrap_errors_normal_xthreat_rerun.csv')
```

```python
# Look at whether the two dataframes have the same values
for col in df_old.columns:
    if col in df_new.columns:
        are_equal = np.allclose(df_old[col], df_new[col])
        print(f"Column '{col}' equal: {are_equal}")
    else:
        print(f"Column '{col}' not found in new dataframe.")
```

Conclusion: the rerun was succesfull and the results are reproducable
